# CrossSection tutorial (Bokeh)

Scaffold examples for `vdapseisutils.core.maps.bokeh.CrossSection`.


```python
import numpy as np
from bokeh.io import output_notebook, show

from vdapseisutils.core.maps.bokeh import CrossSection

output_notebook()
```

### Helpers and volcano inputs

```python
VOLCANOES = {
    "Mount Augustine": {"origin": (59.3626, -153.4333), "azimuth": 75, "radius_km": 35},
    "Mount Rainier": {"origin": (46.8523, -121.7603), "azimuth": 95, "radius_km": 45},
    "Crater Lake": {"origin": (42.9446, -122.1090), "azimuth": 110, "radius_km": 35},
    "Kilauea": {"origin": (19.4210, -155.2870), "azimuth": 45, "radius_km": 40},
    "Yellowstone": {"origin": (44.4280, -110.5885), "azimuth": 70, "radius_km": 70},
}

def synthetic_points(origin, n=200, seed=0):
    rng = np.random.default_rng(seed)
    lat0, lon0 = origin
    lats = lat0 + rng.normal(0, 0.10, size=n)
    lons = lon0 + rng.normal(0, 0.10, size=n)
    depths_km = rng.uniform(0.5, 18.0, size=n)
    return lats, lons, depths_km

def show_cross_section(name, spec, seed=0):
    cs = CrossSection(
        origin=spec["origin"],
        azimuth=spec["azimuth"],
        radius_km=spec["radius_km"],
        depth_extent=(-25, 4),
        figsize=(9, 3.5),
        title=f"{name} - Bokeh Cross Section",
    )
    lats, lons, depths_km = synthetic_points(spec["origin"], seed=seed)
    cs.scatter(
        lat=lats,
        lon=lons,
        z=depths_km,
        z_unit="km",
        z_dir="depth",
        s=20,
        alpha=0.7,
        color="royalblue",
        hover_text=[f"{name} event {i}" for i in range(len(depths_km))],
    )
    show(cs.figure)
```

### Mount Augustine

```python
show_cross_section("Mount Augustine", VOLCANOES["Mount Augustine"], seed=1)
```

### Mount Rainier

```python
show_cross_section("Mount Rainier", VOLCANOES["Mount Rainier"], seed=2)
```

### Crater Lake

```python
show_cross_section("Crater Lake", VOLCANOES["Crater Lake"], seed=3)
```

### Kilauea

```python
show_cross_section("Kilauea", VOLCANOES["Kilauea"], seed=4)
```

### Yellowstone

```python
show_cross_section("Yellowstone", VOLCANOES["Yellowstone"], seed=5)
```
