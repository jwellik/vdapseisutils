### Mount Augustine — simple map

```python
from vdapseisutils.core.maps.bokeh import Map

# Summit ~59.363°N, 153.433°W (USGS AVO)
m = Map(
    origin=(59.3626, -153.4333),
    radial_extent_km=35,
    figsize=(8, 7),
)
m.add_terrain()
show(m.figure)
```

### Regional-style map (Augustine + neighbours)

Same pattern as the Mapping tutorial regional snippet (without needing matplotlib ``Figure`` / ``SubFigure``).

```python
from vdapseisutils.core.maps.bokeh import Map

volc = {
    "name": "Augustine",
    "coords": (59.3626, -153.4333, 1252),
    "map_extent": [-153.95, -152.90, 59.05, 59.70],
    "points": [
        ((59.520, -153.120), "Iliamna"),
        ((59.033, -153.090), "Douglas"),
        ((58.775, -153.670), "Katmai"),
    ],
}

fig_m = Map(map_extent=volc["map_extent"], figsize=(8, 6), title="")
fig_m.add_terrain()
fig_m.set_title("Mount Augustine (Regional View)")

fig_m.plot_volcano(volc["coords"][0], volc["coords"][1], "^r")
[fig_m.plot_peak(p[0][0], p[0][1], "^k") for p in volc["points"]]

show(fig_m.figure)
```

### Kīlauea

Fetch HV stations from IRIS, load the bundled eruption catalog, and mirror the matplotlib mapping workflow. **`show(fig.layout)`** includes the small world locator when ``add_world_location_map()`` is used.

```python
import obspy
from obspy.clients.fdsn import Client

from vdapseisutils.core.maps.bokeh import Map

kilauea = {}
kilauea["coords"] = [19.421, -155.287, 1222]
kilauea["map_extent"] = [-155.500, -154.800, 19.200, 19.500]

client = Client("IRIS")
inv = client.get_stations(
    network="HV",
    minlatitude=kilauea["map_extent"][2],
    maxlatitude=kilauea["map_extent"][3],
    minlongitude=kilauea["map_extent"][0],
    maxlongitude=kilauea["map_extent"][1],
)

cat_kilauea = obspy.read_events("../data/catalog_kilauea_eruption.xml", format="QUAKEML")

fig = Map(map_extent=kilauea["map_extent"], figsize=(10, 4))
fig.add_terrain(verbose=True)
fig.add_scalebar()
fig.plot_inventory(inv)
fig.plot_catalog(cat_kilauea)
fig.add_world_location_map(position="upper right", size=0.25)
fig.set_title("Kīlauea, Hawaiʻi")
fig.set_catalog_subtitle(cat_kilauea)
fig.info()

show(fig.layout)
```
