# Mount Augustine, Alaska (Bokeh)

Simple `vdapseisutils.core.maps.bokeh.Map` with Esri hillshade + Carto overlay (`add_terrain()`), inline in the notebook.

**Setup:** from this repository root run `pip install -e .` (editable install). Then **restart the Jupyter kernel** so Python loads that checkout.

If you see `ModuleNotFoundError: vdapseisutils.core.maps.bokeh`, you are likely importing an older **non-editable** copy from `site-packages`. Fix: `pip uninstall vdapseisutils` (repeat until none left), then `pip install -e .` again from the repo root, restart kernel. Quick check: `import vdapseisutils` → `vdapseisutils.__file__` should point inside your git clone, not only under `.../site-packages/vdapseisutils/` as a flat install.

```python
from bokeh.io import output_notebook, show

from vdapseisutils.core.maps.bokeh import Map

output_notebook()

# Mount Augustine — summit ~59.363°N, 153.433°W (USGS AVO)
m = Map(
    origin=(59.3626, -153.4333),
    radial_extent_km=35,
    figsize=(8, 7),
)
m.add_terrain()
show(m.figure)
```
