# Mount Augustine, Alaska (Bokeh)

Simple `vdapseisutils.core.maps.bokeh.Map` with Esri hillshade + Carto overlay (`add_terrain()`), inline in the notebook.

```python
from bokeh.io import output_notebook, show

from vdapseisutils.core.maps.bokeh import Map

output_notebook()

# Mount Augustine — summit ~59.363°N, 153.433°W (USGS AVO)
m = Map(
    origin=(59.3626, -153.4333),
    radial_extent_km=35,
    figsize=(8, 7),
)
m.add_terrain()
show(m.figure)
```
