If `ModuleNotFoundError: vdapseisutils.core.maps.bokeh` persists, Python is still picking up an **old tree** under `site-packages` (missing the `core/maps/bokeh` package).

**When `pip uninstall vdapseisutils` says *Can't uninstall … No files were found*** — pip’s metadata is out of sync (common with **conda + pip**). Fix the environment, then reinstall:

1. `conda list vdapseisutils` — if it appears, run **`conda remove vdapseisutils`** (that removes the conda-managed copy).
2. Still broken: in that conda env’s **`site-packages`**, manually remove **`vdapseisutils/`** (the package directory), **`vdapseisutils-*.dist-info/`**, and any **`__editable___vdapseisutils*.pth`** / stray `.egg-link` pointing at an old checkout.
3. From the **repo root**: **`pip install -e .`**, then **restart the kernel**.

After a good editable install, **`import vdapseisutils; vdapseisutils.__file__`** should list a path **inside your git clone** (or an `__editable__` redirect to it), not a standalone old copy only under `site-packages`.

**More notes from real installs**

- **`conda remove vdapseisutils`** → **`PackagesNotFoundError`**: conda never installed this project; that is **normal**. Skip conda and use steps 2–3 (remove leftover dirs under `site-packages`, then **`pip install -e .`**).
- While **`pip install -e .`** runs, you may still see **`Can't uninstall 'vdapseisutils'. No files were found`** — pip is bumping into **orphan metadata**. Ignore it if the run ends with **`Successfully installed vdapseisutils-0.2.0`** (same version as `[project]` in `pyproject.toml`).

```python
from bokeh.io import output_notebook, show

from vdapseisutils.core.maps.bokeh import Map

output_notebook()

# Mount Augustine — summit ~59.363°N, 153.433°W (USGS AVO)
m = Map(
    origin=(59.3626, -153.4333),
    radial_extent_km=35,
    figsize=(8, 7),
)
m.add_terrain()
show(m.figure)
```
