## Regional-style example (Mapping tutorial pattern)

This mirrors the `Mapping_tutorial.ipynb` regional map pattern using Bokeh `Map` (without `add_world_location_map()` and `add_scalebar()` for now).

```python
# Regional Map (Bokeh adaptation)
from bokeh.io import show

volc = {
    "name": "Augustine",
    "coords": (59.3626, -153.4333, 1252),
    "map_extent": [-153.95, -152.90, 59.05, 59.70],
    "points": [
        ((59.520, -153.120), "Iliamna"),
        ((59.033, -153.090), "Douglas"),
        ((58.775, -153.670), "Katmai"),
    ],
}

fig_m = Map(map_extent=volc["map_extent"], figsize=(8, 6), title="")
fig_m.add_terrain()
fig_m.figure.title.text = "Mount Augustine (Regional View)"

fig_m.plot_volcano(volc["coords"][0], volc["coords"][1], "^r")
[fig_m.plot_peak(p[0][0], p[0][1], "^k") for p in volc["points"]]

show(fig_m.figure)
```

```python
from bokeh.io import output_notebook, show

from vdapseisutils.core.maps.bokeh import Map

output_notebook()

# Mount Augustine — summit ~59.363°N, 153.433°W (USGS AVO)
m = Map(
    origin=(59.3626, -153.4333),
    radial_extent_km=35,
    figsize=(8, 7),
)
m.add_terrain()
show(m.figure)
```
